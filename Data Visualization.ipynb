```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df=pd.read_csv("housing.csv")

df["income_cat"] = pd.cut(df["median_income"],
                             bins=[0, 1.5, 3.0, 4.5, 6.0, np.inf],
                             labels=[1, 2, 3, 4, 5])
```

```python
df
```

```python
from sklearn.model_selection import StratifiedShuffleSplit

# Assume income_cat is a column in the dataset created from median_income
split = StratifiedShuffleSplit(n_splits=1, test_size=0.2, random_state=42)

for train_index, test_index in split.split(df, df["income_cat"]):
    strat_train_set = df.loc[train_index]
    strat_test_set = df.loc[test_index]
```

```python
strat_train_set 
```

```python
strat_test_set 
```

```python
for sett in (strat_test_set,strat_train_set):
    sett.drop("income_cat",inplace=True,axis=1)
```

```python
df=strat_train_set.copy()
```

```python
df
```

```python
df.plot(kind="scatter",x="longitude",y="latitude",grid=True,color="green")
plt.title("House density distribution")
```

```python
df.plot(kind="scatter",x="longitude",y="latitude",grid=True,color="green",alpha=0.3)
```

```python
df.plot(kind="scatter",x="longitude",y="latitude",grid=True,cmap="jet",c="median_income")
plt.title("median income distribution")
df.plot(kind="scatter",x="longitude",y="latitude",grid=True,alpha=0.9,cmap="jet",c="median_house_value")
plt.title("median house value")

```
