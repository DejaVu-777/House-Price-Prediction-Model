Below i'm trying to split the whole dataset into testing and training datasets .I wrote a simple python programe which takes the parameters of the dataset and the test ration split.It then calculates the len of the data multiplied with the test_ratio.training and testing is split then .

```python
def shuffle_split(data,test_ratio):
    random.seed(42)
    shuffled_indices=np.random.permutation(len(data))
    #print(shuffled_indices)
    test_set_size = int(len(data) * test_ratio)
    training_set=shuffled_indices[:test_set_size]
    test_set=shuffled_indices[test_set_size:]
    
    return data.iloc[ training_set] , data.iloc[test_set]
    
```
